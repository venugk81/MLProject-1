```python
def tokenize(sentence):
    sentence = sentence.lower() 
    sentence = re.sub(r'[^\w\s]', '', sentence)        # \w is for A-Z,a-z,0-9, _, \s is for space, ^ - not condition
    return sentence.split()
```

```python
# string = 'I #like pizza!! I enjoy @eating it. I will go for having the pizza at 3 p.m. today.'
# re.sub(r'[^\w\s]', '', string.lower()).split()
```

```python
tokenized_sentences = [tokenize(t) for t in text]
tokenized_sentences[:5]
```

```python
def build_vocabulary(sentences):
    vocabulary = set()
    for word in sentences:
        vocabulary.update(word)
    return sorted(list(vocabulary))
```

```python
vocabulary = build_vocabulary(tokenized_sentences)
vocabulary
```

```python
len(vocabulary)
```

```python
def create_bow_vector(sentence, vocabulary): 
    vec = [0]*len(vocabulary)
    for word in sentence:
        if word in vocabulary:
            idx = vocabulary.index(word)
            vec[idx] = 1
    return vec
```

```python
bow_vectors = [create_bow_vector(sentence, vocabulary) for sentence in tokenized_sentences]
# bow_vectors[0]
```

```python
len(bow_vectors)
```

```python
encoded_vectors = np.array(bow_vectors)
encoded_vectors[0]
```

```python
cols = [i for i in vocabulary]
```

```python
df[cols] = encoded_vectors
```

### Using sk-learn

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
texts = [text for text, label in dataset]
labels = [label for text, label in dataset]
```

```python
vectorizer = CountVectorizer()
```

```python
bow_matrix = vectorizer.fit_transform(texts)
```

```python
bow_matrix
```

```python
bow_matrix.toarray()
```

```python
vectorizer.get_feature_names_out()
```

```python
bow_df = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
bow_df['label'] = labels
```
