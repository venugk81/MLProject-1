## TF-IDF

#### Term Frequency (TF)

```python
import pandas as pd
import numpy as np
import re
```

```python
corpus = [
    "I love to eat pizza because pizza is tasty",
    "Pizza is very delicious",
    "I hate cold pizza"
]
```

### Scratch Implementation

```python
def tokenize(sentence):
    sentence = sentence.lower()
    sentence = re.sub(r'[^\w\s]', '', sentence)
    return sentence.split()
```

Calculating TF

```python
def get_term_frequency(corpus):

    vocabulary = []
    for c in corpus:
        splitted_sentence = tokenize(c)
        for word in splitted_sentence:
            vocabulary.append(word)

    vocabulary = set(vocabulary)

    term_freq = []
    for sentence in corpus:
        splitted_sentence = tokenize(sentence)
        vocab_dict = {v:0 for v in vocabulary}
        
        for word in splitted_sentence:
            vocab_dict[word] += 1
            
        for v in vocabulary:
            tf = vocab_dict[v]/len(splitted_sentence)
            vocab_dict[v] = tf
            
        term_freq.append(vocab_dict)

    return term_freq
```

```python
term_freq = get_term_frequency(corpus)
term_freq
```

```python
df = pd.DataFrame(term_freq)
df.index = corpus

display(df)
```

#### Inverse Document Frequency (IDF)

```python
def get_inverse_document_frequency(corpus):
    vocabulary = []
    for c in corpus:
        sentence = tokenize(c)
        for word in sentence:
            vocabulary.append(word)

    vocabulary = set(vocabulary)

    total_documents = len(corpus)

    inverse_document_frequency = {}
    for v in vocabulary:
        num_docs = 0           # number of documents containing a word
        for sentence in corpus:
            splitted_sentence = tokenize(sentence)
            if v in splitted_sentence:
                num_docs += 1
        inverse_document_frequency[v] = np.log(total_documents/num_docs)
        
    return inverse_document_frequency
```

```python
inverse_document_frequency = get_inverse_document_frequency(corpus)
inverse_document_frequency
```

Calculating TF-IDF

```python
def get_tf_idf(corpus):
    tf_list = get_term_frequency(corpus)
    idf = get_inverse_document_frequency(corpus)

    tf_idf = []
    for tf_dict in tf_list:
        tf_idf_sentence = {}
        for word, term_freq in tf_dict.items():
            tf_idf_sentence[word] = term_freq * idf[word]
        tf_idf.append(tf_idf_sentence)

    return tf_idf
```

```python
tf_idf = get_tf_idf(corpus)
tf_idf
```

```python
df = pd.DataFrame(tf_idf)
df.index = corpus
display(df)
```

### Using sk-learn

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf_idf_vectorizer = TfidfVectorizer()
```

```python
tf_idf_rep = tf_idf_vectorizer.fit_transform(corpus).todense()
```

```python
tf_idf_rep
```

```python
df = pd.DataFrame(tf_idf_rep)
df.columns = tf_idf_vectorizer.get_feature_names_out()
df.index = corpus
display(df)
```
