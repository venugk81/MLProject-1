## Email Classification - Spam or Ham Detection

```python
import pandas as pd
```

```python
# Reading the dataset
df = pd.read_csv('Email_dataset.csv', index_col=0)
```

```python
df.head()
```

```python
df.info()
```

```python
Email = df['Email']
Email
```

#### Applying BOW to encode the text data

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer()
```

```python
bow_matrix = vectorizer.fit_transform(Email)
bow_matrix
```

```python
bow_matrix.toarray()
```

```python
vectorizer.get_feature_names_out()
```

#### Creating a new dataframe after text encoding

```python
bow_df = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out())
```

```python
bow_df.head()
```

```python
bow_df.shape
```

#### Adding a label column

```python
bow_df['Label'] = df['Label'].apply(lambda x:0 if x=='ham' else 1)
```

```python
bow_df['Label'].head()
```

```python
bow_df.shape
```

```python
bow_df.head()
```

#### Dividing the dataset into X and y

```python
X = bow_df.drop('Label', axis=1)
y = bow_df['Label']
```

#### Splitting the dataset into train and test data

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

#### Model Building

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
print('Training Score:', model.score(X_train,y_train))
```

```python
print('Test Score:', model.score(X_test,y_test))
```

#### Observation:
- The model shows a training score of 99%
- It shows a test score of 97%
