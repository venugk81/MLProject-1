```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

#### Barplot using Seaborn on Tips Dataset

```python
sns.barplot(x="day", y="total_bill", data=df)
plt.show()
```

Ques. Calculate the average tip by day.

```python
sns.barplot(x='day', y='tip', data=df, color='skyblue')
plt.title('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

#### Barplot on horizontal axis

```python
sns.barplot(x='tip', y='day', data=df, color='skyblue')
plt.title('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

This black line is known as Margin of Error.

### KDE Plot

A KDE (Kernel Density Estimate) plot is a smooth curve that shows the probability density of a continuous variable. It basically estimates the distribution of the data.

```python
np.random.seed(42)
x = np.random.randint(1, 50, size=100)
x
```

```python
sns.kdeplot(x)
plt.show()
```

```python
sns.kdeplot(x='total_bill', data=df, color='red')
plt.show()
```

```python
sns.kdeplot(x='tip', data=df, color='green')
plt.show()
```

### Boxen Plot
- It is an advanced version of a box plot, designed to visualize large datasets 
- It goes further by showing additional layers of data beyond the quartiles, especially in the tails where the data is less dense.
- It is also known as a Letter Value Plot.

```python
# draw boxenplot on tips dataset
sns.boxenplot(y='total_bill', data=df)
plt.show()
```

```python
sns.boxenplot(y='tip', data=df)
plt.show()
```
