### Datetime Functions
These functions help in parsing, manipulating, and analyzing datetime values.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Ecomm_Sales.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
df.head()
```

```python
df.dtypes
```

```python
row = df['order_date'][0]
row
```

```python
row.year                 # prints the year from the date
```

```python
row.month                # prints the month from the date
```

```python
row.day                  # prints the day from the date
```

```python
row.hour                 # prints the hour
```

```python
row.minute               # prints the minute
```

```python
row.second               # prints the second
```

#### strftime (stringformat time)
It is used to modify the datetime format

```python
row = df['order_date'][0]
row
```

```python
row.strftime('%Y')
```

```python
row.strftime('%m')
```

```python
row.strftime('%d')
```

```python
row.strftime('%m-%d')
```

```python
row.strftime('%d/%m/%Y')
```

```python
row.strftime('%B')        # %B --> displays the month name
```

```python
df['order_date'][100].strftime('%B')
```

A comprehensive list of other formats can be found here: <br>
https://pandas.pydata.org/docs/reference/api/pandas.Period.strftime.html

Thus, strftime() is used to display datetime information in a specific, human-readable format, especially when generating reports, creating user-friendly visualizations, or exporting data.

```python
df['order_date'][100].strftime('%A')
```
