#### Setting-Resetting Index

- There are implicit indices for both rows and columns.
- Labels provided to columns are explicit index.
- If we want to set any column as an index to the dataframe, we can use set_index method.

```python
# display index before setting
df.index
```

```python
# Set index
df.set_index('Id', inplace=True)
```

```python
df.head()
```

```python
# Here explicit index is used for both rows and columns
df.loc[2:4,'Sepal Length':'Petal Length']
```

```python
df.index
```

```python
# for resetting the index back, we use reset method
df.reset_index(inplace=True)
```

```python
df.head()
```

#### Sorting Data in Pandas

```python
# by default it sorts the data in ascending order
df.sort_values('Sepal Length (in cm)')
```

```python
df.sort_values('Sepal Width (in cm)')
```

For sorting the data in descending/decreasing order, we set ascending=False parameter

```python
df.sort_values('Sepal Width (in cm)', ascending=False)
```

Sorting Multiple columns together

```python
df.sort_values('Sepal Length (in cm)')
```

```python
df.sort_values(['Sepal Length (in cm)','Sepal Width (in cm)'])
```

```python
# Sorting multiple columns in decreasing order
df.sort_values(['Sepal Length (in cm)','Sepal Width (in cm)'], ascending=False)
```

```python
# Sorting multiple columns in different orders
df.sort_values(['Sepal Length (in cm)', 'Sepal Width (in cm)'], ascending=[False, True])
```

#### Masking

```python
df.head()
```

```python
# Ques. Pick all rows with petal length > 2
df['Petal Length (in cm)'] > 2
```

```python
# For viewing the data, we apply masking
df[df['Petal Length (in cm)'] > 2]
```

```python
# We can even display a single column also after masking
# df['Petal Length (in cm)'][df['Petal Length (in cm)'] > 2]
# or
df[df['Petal Length (in cm)'] > 2]['Petal Length (in cm)']
```

```python
# Applying Masking for multiple conditions
# Display all rows having Petal Length > 2 and < 4

mask = (df['Petal Length (in cm)'] > 2) & (df['Petal Length (in cm)'] < 4)
```

```python
df[mask]
```

```python
# Ques. Display the row having Sepal Length as 7.0
df[df['Sepal Length (in cm)'] == 7.0]
```
