#### nan and None Type -

#### None -
1. In Python, None is used to indicate the absence of a value or a "null" state.
2. None is of type NoneType.
3. In NumPy arrays, using None will automatically convert the array's data type to object.

```python
type(None)
```

```python
a = np.array([3,7,1,8,None])
a
```

#### nan -
1. nan stands for "Not a Number".
2. nan was introduced to handle missing or undefined values specifically in the context of numerical computations.
3. It is a floating-point value, so it can only be used with arrays of floating-point type.

```python
type(np.nan)
```

```python
b = np.array([1,2,3,4,np.nan])
b
```

3. Let Survey score greater than 6 is known as promoter score. Display the promoter score.
4. Let Survey score less than 6 is known as detractor score. Display the detractor score.
5. Calculate the promoter and detractor percentage.

```python
print('Length of the array:', len(score))
```

```python
min_score = np.min(score)
print('Minimum Score:',min_score)
```

```python
max_score = np.max(score)
print('Maximum Score:',max_score)
```

#### np.where() method

It used to locate the indices of elements in an array that satisfy a given condition

```python
arr = np.array([13,-37, 20, 12, -10, -19, 28, 40, -26, 32])
arr
```

```python
np.where(arr<0)
```

```python
np.where(arr<20)
```

Also, np.where can be used with additional parameters to act more like a conditional replacement operation.<br>
np.where(condn, truevalue, falsevalue)

```python
np.where(arr<0, '-ve', '+ve')
```

```python
arr = np.array([4, 19, 15, 31, 40, 26, 35, 5, 18, 43, 10])
arr
```

```python
np.where(arr%5==0)
```

If not providing condition, it returns the index of all non-zero numbers

```python
arr2 = np.array([2,3,0,8,0,9,1,4,5,0,7])
np.where(arr2)
```

#### Iteration using np.nditer()

- It provides advanced iteration with element-wise access and modification capabilities.
- It is ideal for high-performance tasks requiring control.

```python
a = np.array([3,4,5,6,7])
```

```python
# Element-wise iteration using nditer
for i in np.nditer(a):
    print(i)
```

```python
arr = np.array([[10, 20, 30],[40, 50, 60]])
arr
```

```python
arr.shape
```

```python
for i in np.nditer(arr):
    print(i)
```

#### Sorting Numpy Arrays

```python
a = np.array([23, 65, 10, 45, 2, 17, 38, 19, 44, 26])
```

```python
a
```

```python
np.sort(a)          # this does not make a permanent change in the original array
```

```python
a
```

```python
a.sort()          # this makes a permanent change in the original array
```

```python
a
```

**Numpy, Pandas:**<br>
- When any method returns an output, then it does not make a permanent change.<br>
- When any does not returns an output, it makes a permanent change to the original dataset.

Sorting a 2D array

```python
x = np.array([[20,4,8],[12,2,10]])
x
```

```python
# by default it sorts the array row-wise
np.sort(x)
```

```python
np.sort(x, axis=0)     # in numpy, axis=0 is used with columns
```

```python
np.sort(x, axis=1)     # in numpy, axis=1 is used with rows
```
