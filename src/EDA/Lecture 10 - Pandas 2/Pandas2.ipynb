#### Indexing and Slicing in Pandas

##### loc/iloc methods

There are two types of indexing-
1. Implicit Index (Internal indexing)
2. Explicit Index

There is only one type of internal indexing in Python and that starts from 0.
In Pandas too, implicit index is applied by default but if we change the index it is known as explicit indexing.

```python
# iloc - it is an integer location based indexing. It takes the default index even though explicit index is applied
# syntax - df.iloc[row,col] 

# Accessing the element of first row & first column
df.iloc[0,0]
```

```python
# Accessing the entire first row
df.iloc[0]
```

```python
df.iloc[2,1]
```

```python
# Use slicing to print a range of elements
df.iloc[0:5,0]
```

```python
df.iloc[0:5,1]
```

```python
df.iloc[0:7,5]
```

```python
# Accessing the first row and multiple columns
df.iloc[0,2:5]
```

```python
df.iloc[0:2, 3:5]
```

```python
df.iloc[2:6,1:4]
```

```python
# List of indices can also be used to display random rows and columns
df.iloc[[0,10,54,7,28]]
```

```python
# df.iloc[row, column]
df.iloc[[0,100,25,65],[2,5,1]]
```

```python
df.iloc[[0,6,9,3],[0,5,3]]
```

```python
# loc - it is a label-based indexing. It uses rows and columns based on the labels provided
# syntax - df.loc[row,col]

# Accessing first row
df.loc[0]
```

```python
# Accessing last row
df.loc[149] 
```

```python
# Accessing specific Column of a specific row
df.loc[0,"Species"]
```

```python
df.loc[1,"Sepal Length"]
```

```python
df.loc[4,'Sepal Width']
```

```python
# Accessing multiple rows of a single column using list of indices
df.loc[[149,23,0,74], "Species"]
```

```python
df.loc[[149,23,0,74], "Sepal Length"]
```

```python
# here too list of indices can also be passed to display random rows and columns
df.loc[[100, 60, 12, 35],['Species', 'Sepal Length', 'Petal Length']]
```

```python
df.loc[0:5, 'Petal Length':'Species']
```

#### Missing Value Detection using Pandas
- A missing value signifies an empty cell.
- Missing values are detected using isnull() or isna() methods

```python
df.isnull()
```

```python
df.isna()
```

```python
# Verifying existance of Null values column wise
df.isnull().any()
```

```python
# column wise count of null values
df.isnull().sum()
```

As we can see that Sepal Length column has a missing, let us apply masking to print the row with missing value

```python
df["Sepal Length"].isna().sum()
```

```python
df[df["Sepal Length"].isna() == True]           # this is masking
```

Similarly null values can be checked for other columns as well

```python
df[df["Sepal Width"].isna() == True]
```

```python
df[df['Petal Length'].isna() == True]
```

**Removing nulls using dropna()**

```python
df.dropna()
```

```python
df.isna().sum()
```

Changes are not observed in the original dataframe. Hence we use inplace=True for making the permanent change in the original dataset

```python
# this drops all the NaNs from original dataframe
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
df.shape
```

**But what is the problem with deleting rows/columns?**
- loss of valuable data<br>
So instead of dropping, it is better to fill the missing values with some data using **fillna()** method.
