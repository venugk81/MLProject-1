#### Stacking 
Stacking involves placing arrays on top of each other or alongside each other to create a new array with an additional dimension.

```python
array1 = np.array([1,2,3,4,5])
array1
```

```python
array2 = np.array([5,6,7,8,9])
array2
```

```python
array3 = np.stack((array1, array2))
array3
```

Two types of stacking methods:
1. hstack - combining two arrays horizontally
2. vstack - combining two arrays vertically

```python
np.hstack([array1,array2])
```

```python
np.vstack([array1,array2])
```

1. If there are equal number of rows then hstack is applied
2. If there are equal number of columns then vstack is applied

```python
m = np.arange(1,4)
m
```

```python
n = np.arange(0,2)
n
```

```python
o = np.arange(1,8)
o
```

```python
np.hstack([m,n,o])
```

#### Concatenation
- Joins arrays along an existing axis
- Does not add a new dimension

```python
a = np.arange(2)
a
```

```python
b = np.arange(4, 8)
b
```

```python
c = np.concatenate((a, b))
c
```

Concatenating 2D arrays

```python
# Concatenate along axis=0
a = np.arange(6).reshape(2, 3)
a
```

```python
b = np.arange(6, 12).reshape(2, 3)
b
```

```python
c = np.concatenate((a, b), axis=0)
c
```

```python
# Concatenate along axis=1
c = np.concatenate((a, b), axis=1)
c
```

#### Broadcasting
A process that allows arrays with different shapes to be used in arithmetic operations.<br>
It stretches or replicate smaller arrays to match the dimensions of larger ones during these operations.

```python
a = np.arange(1,13).reshape(-1,3)
a
```

```python
b = np.array([8, 1, 6])
b
```

```python
a*b
```

Here array b is broadcasted<br>
np.tile() function do broadcasting as it repeats rows or columns many times

```python
a = np.arange(1,5)
a
```

```python
b = np.arange(5,9).reshape(4,1)
b
```

```python
a+b
```

a has shape 1x4 and b has shape 4x1 then a+b will have shape (any mathematical operation) 4x4 as both a and b are broadcasted.

#### Generating random numbers using numpy

**np.random.rand()** - generate random floating-point numbers between 0 (inclusive) and 1 (exclusive)

```python
# Generate 1D random numbers
random_1D = np.random.rand(5)
random_1D
```

```python
# Generate 2D random numbers
random_2D = np.random.rand(3, 4)
random_2D
```

**np.random.uniform()** - generate random numbers that are uniformly distributed over a specified range

```python
random_uniform = np.random.uniform(10, 20, size=(2, 3))
random_uniform
```

**np.random.normal()** - generates samples from a normal distribution with specified mean and standard deviation

```python
random_normal = np.random.normal(loc=5, scale=2, size=(3, 4))
random_normal
```

- loc → Mean (μ) of the normal distribution
- scale → Standard deviation (σ) of the normal distribution
- size → Shape of the output

**np.random.randint()** - generates random integers within a specified range

```python
random_integers = np.random.randint(10, 20, size=(2, 3))
random_integers
```

**np.random.choice()** - Random selection of elements

**Reproducible Randomness:**
- Reproducibility is important in projects where results must remain consistent, such as in scientific experiments or debugging
- To make random numbers reproducible, we use the np.random.seed() function
- Setting a seed ensures the same sequence of random numbers is generated every time we run the code

```python
# Running the code twice
random_numbers_1 = np.random.rand(3)
print("First Run:", random_numbers_1)
```

```python
# Running the code twice
random_numbers_2 = np.random.rand(3)
print("Second Run:", random_numbers_2)
```

The two outputs differ

```python
np.random.seed(42)
print(np.random.rand(3))
```

```python
np.random.seed(42)
print(np.random.rand(3))
```

```python
np.random.seed(42)
print(np.random.rand(6))
```

Seed is an initial value given to the random number generator to produce a reproducible sequence of random numbers.

**Note: random_state ---> used in ML & DL for maintaining consistency**
