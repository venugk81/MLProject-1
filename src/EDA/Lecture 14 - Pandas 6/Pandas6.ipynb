Ques. Create a new column bill_category and check if:
1. Bill < 10 - Low
2. Bill between 10–20 - Medium
3. Bill > 20 - High

```python
def bill_cat(row):
    if row < 10:
        return 'Low'
    elif row > 20:
        return 'High'
    else:
        return 'Medium'
```

```python
df['bill_category'] = df['total_bill'].apply(bill_cat)
```

```python
df.head()
```

Ques. Calculate the total bill of each row after adding a 5% service charge. Store it in a new column called final_bill.

```python
# how to calculate 5% of a value
# 5/100 ----> 0.05
# x = 200
# 5% of 200 = 0.05 * 200
#           = 10
```

```python
def service_charge(row):
    service = row * 0.05       # 5% of total bill
    final = service + row
    return final
```

```python
df['final_bill'] = df['total_bill'].apply(service_charge)
```

```python
df.head()
```

Ques. Create a new column called family_visit. Classify each visit as "Family" when the group size is 4 or more, and "Regular" when the group size is less than 4.

```python
def visit(row):
    if row >= 4:
        return 'Family'
    else:
        return 'Regular'
```

```python
df['family_visit'] = df['size'].apply(visit)
```

```python
df.head()
```

#### Rolling Function
- It is used to perform moving window operations on numerical data (current row and the previous n rows).
- It allows us to apply functions like mean, sum, standard deviation, etc. over a sliding window of values.

**Syntax -** df['column'].rolling(window).function()<br>
window - no. of rows to consider<br>
function - can be any function

```python
df['tip'].head()
```

```python
df['tip'].rolling(2).sum()              # rolling is specific to time series data
```

```python
df['tip'].rolling(5).min()
```

```python
df['total_bill'].rolling(2).mean()
```

```python
df['total_bill'].rolling(4).median()
```

#### Expanding
- This method is similar to rolling(), but instead of using a fixed-size moving window, it uses all the data from the beginning up to the current row.
- This is useful for cumulative calculations.

```python
df['total_bill'].expanding().sum()
```

```python
df['tip'].expanding().mean()
```

```python
df['final_bill'].expanding().min()
```

#### Nested Data
- For nested JSON structures, you can normalize the data for flat tabular representation using json_normalize

```python
from pandas import json_normalize
```

```python
data = [
            {
                "id": 1,
                "name": "Rahul",
                "department": "IT",
                "subjects": [
                                {"subject": "Python", "marks": 90},
                                {"subject": "SQL", "marks": 85}
                            ]
            },
            {
                "id": 2,
                "name": "Aman",
                "department": "AI",
                "subjects": [
                                {"subject": "Machine Learning", "marks": 95},
                                {"subject": "Deep Learning", "marks": 92}
                            ]
            }
      ]
```

```python
data
```

Without json_normalize

```python
df = pd.DataFrame(data)
df
```

Therefore, we use json_normalize:
- record_path: convert nested data into rows
- meta: for using parent columns

```python
df = json_normalize(data, record_path='subjects', meta=['id', 'name', 'department'])
```
