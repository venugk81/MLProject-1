```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('diamonds')
df.head()
```

#### Heatmap
- It is used to represents the correlation, distribution and patterns in the data.
- As it finds the correlation, hence we create a numerical data

```python
df_numeric = df.select_dtypes(include=['int', 'float'])
```

```python
df_numeric.head()
```

Finding the correlation between all numeric columns

```python
df_numeric.corr()
```

- It is difficult to observe the correlation values using the above table.
- For simplifying, we use a heatmap

```python
sns.heatmap(df_numeric.corr())
plt.show()
```

- The side bar in a heatmap is called a color bar (or legend). It indicates the scale of values represented by the colors in the heatmap.
- For viewing the correlation strength values, we use annot=True.

```python
sns.heatmap(df_numeric.corr(), annot=True)
plt.show()
```

Using different color palettes

```python
print(plt.colormaps())                    # this fetches all the color palettes provided by the library
```

```python
plt.figure(figsize=(10,7))
sns.heatmap(df_numeric.corr(), annot=True, cmap='Pastel1')
plt.show()
```

### Pairplot
It is used for exploring distributions and correlations among multiple numerical features

```python
df.head()
```

```python
sns.pairplot(data=df)
plt.show()
```

Checking the distribution of a single plot

```python
sns.scatterplot(x='depth', y='price', data=df_numeric)
plt.show()
```

1. Each pair of variables (features) is plotted against each other as a scatter plot in individual grid cells. This helps identify potential relationships and correlations between variables.
2. The diagonal cells typically display histograms or kernel density plots of individual variables to show the distribution of each feature with itself.

#### Violin Plot
It is a combination of box plot and kde plot. It is used to show the distribution of numerical data.

```python
plt.violinplot(df['price'])
plt.show()
```

Using Seaborn

```python
sns.violinplot(y='price', data=df)
plt.show()
```

Cross Checking the distribution using Boxplot

```python
sns.boxplot(y=df['price'])
plt.show()
```

#### Stacked Plot

```python
# Ex-
data1 = {
    'Category': ['A', 'B', 'C', 'D'],
    'Value1': [3, 4, 2, 5],
    'Value2': [1, 2, 3, 4],
    'Value3': [2, 1, 5, 3]
}

df1 = pd.DataFrame(data1)
df1
```

```python
df1.plot(kind='bar', stacked=True)
plt.show()
```

```python
dummy = df[['cut','x','y','z']].head(5)
dummy
```

```python
dummy.plot(kind='bar', stacked=True, color=sns.color_palette('Set2'))
plt.show()
```

For changing the position of the legend, we use loc parameter

```python
dummy.plot(kind='bar', stacked=True, color=sns.color_palette('Set2'))
plt.legend(loc=(1,0))                 # loc=(x,y)
plt.show()
```
