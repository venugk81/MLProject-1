### Seaborn

Same kind of plots like matplotlib can be created using seaborn as well.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Lineplot

```python
numList1 = [1,2,3,4,5,6]
numList2 = [5,6,4,7,6,8]

# matplotlib - plt.plot(x,y)
sns.lineplot(x=numList1, y=numList2, marker='o')
plt.show()
```

Matplotlib methods can be used along with seaborn too

#### Bar Plot

```python
students = ['A', 'B', 'C', 'D', 'E']
marks = [67, 89, 54, 72, 83]

# matplotlib - plt.bar(x,y)
sns.barplot(x=students, y=marks)
plt.show()
```

### Histogram
It is a type of a bar graph that represents the distribution of numerical data. It shows how often values (or ranges of values, called bins) occur in a dataset.

**Example:**

Imagine on a survey of 100 people the results of their ages be like this:
- 0–10 years: 5 people
- 10–20 years: 12 people
- 20–30 years: 22 people
- 30–40 years: 30 people
- 40–50 years: 20 people
- 50–60 years: 11 people

A histogram of this data would have bars representing these age ranges, with the height of each bar showing how many people fall into each age group.

```python
import numpy as np
np.random.seed(0)
x = np.random.randint(1, 100, size=100)
```

```python
plt.hist(x, bins=10)
plt.show()
```

```python
plt.hist(x, bins=10, edgecolor='black')      # edgecolor is used for separating out the bins
plt.show()
```

```python
plt.hist(x, bins=10, edgecolor='black')
plt.show()
```

#### Histogram using Seaborn 
It offers better visualization

```python
sns.histplot(x, bins=10)
plt.show()
```

```python
sns.histplot(x, bins=10, kde=True)           # kde ---> used for density estimation
plt.show()
```

#### Box Plot

It is a graph that shows the spread and distribution of a dataset based on five summary statistics:  
- Minimum
- First Quartile (Q1) - 25%
- Median (Q2) - 50%
- Third Quartile (Q3) - 75%
- Maximum
<br>

- A box plot helps you visualize the range, center, and variability of data.
- It is used for identifying outliers

#### How it works:
1. Box: Represents the middle 50% of the data (from 25% to 75%).  
2. Line inside the box: The median (middle value) of the dataset.
3. Whiskers: Lines that extend from the box to the minimum and maximum values (excluding outliers).
4. Dots or stars outside the whiskers: These are outliers or unusually high or low values.

```python
np.random.seed(0)
x = np.random.randint(1, 100, size=100)
```

```python
plt.boxplot(x)
plt.show()
```

```python
sns.boxplot(y='total_bill', data=df)
plt.show()
```

```python
sns.boxplot(y='tip', data=df, color='blue')
plt.show()
```

### Pie Chart
It is a circular graph used to represent parts of a whole. Each slice of the pie shows a category’s proportion or percentage of the total.

```python
venue = ['Hyderabad', 'Mumbai', 'Chennai', 'Delhi', 'Kolkata']
matches = [20, 5, 19, 11, 32]

plt.pie(matches, labels=venue)
plt.show()
```

```python
venue = ['Hyderabad', 'Mumbai', 'Chennai', 'Delhi', 'Kolkata']
matches = [20, 5, 19, 11, 32]

plt.pie(matches, labels=venue, autopct="%0.2f%%")
plt.show()
```

```python
# adding colour
colors = ['pink', 'lightgreen', 'skyblue', 'lightyellow', 'grey']

venue = ['Hyderabad', 'Mumbai', 'Chennai', 'Delhi', 'Kolkata']
matches = [20, 5, 19, 11, 32]

plt.pie(matches, labels=venue, 
        autopct='%0.2f%%', 
        colors = colors)

plt.show()
```

Seaborn does not have a native function to create pie charts

```python
# using color palette

plt.pie(matches, labels=venue,
        autopct='%0.2f%%', 
        colors = sns.color_palette("Set2"), 
       )
plt.show()
```

```python
# startangle
plt.pie(matches, labels=venue,
        autopct='%0.2f%%', 
        colors = sns.color_palette("Set2"), 
        startangle = 90,
       )
plt.show()
```

```python
# startangle
plt.pie(matches, labels=venue,
        autopct='%0.2f%%', 
        colors = sns.color_palette("Set2"), 
        startangle = 0,              # default is 0
       )
plt.show()
```

```python
# explode
explode = [0, 0.4, 0, 0, 0]

plt.pie(matches, labels=venue,
        autopct='%0.2f%%', 
        colors = sns.color_palette("Set2"), 
        explode = explode,
       )
plt.show()
```

```python
# explode on multiple values
explode = [0, 0.3, 0, 0.2, 0]

plt.pie(matches, labels=venue,
        autopct='%0.2f%%', 
        colors = sns.color_palette("Set2"), 
        explode = explode,
       )
plt.show()
```

#### Donut Chart
A donut chart is simply a pie chart with a hole. The purpose of the hole is not to change the data; it is to improve readability and provide space for additional information such as totals, percentages, etc.

```python
fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
data = [35, 25, 20, 20]
total = sum(data)

plt.pie(data, labels = fruits, autopct='%1.1f%%', 
        startangle = 90, 
        pctdistance = 0.85,                                     # pctdistance decides how far from the center the percentages should be
        wedgeprops = {"width": 0.4,"edgecolor": "white"},       # width = 0.4 makes the hole
        colors = sns.color_palette('Set2'),
       )

plt.text(0, 0, f'Total\n{total}', ha='center', va='center',fontsize=12)

plt.title("Fruit Distribution Donut Chart")
plt.show()
```
