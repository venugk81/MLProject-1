### Advance Statistical testing using scipy.stats functions

#### 1. Chi-Square Test for Goodness-of-Fit
- How to Calculate: Use `scipy.stats.chisquare(f_obs, f_exp)`.
- Null Hypothesis ($H_0$): The observed proportions of males and females match the expected proportions (50-50 split).
- Alternative Hypothesis ($H_a$): The observed proportions of males and females do not match the expected proportions.


```python
from seaborn import load_dataset
from scipy import stats
```

```python
# Load dataset and define population
tips = load_dataset("tips")
sample = tips['sex']
observed = sample.value_counts().values
expected = [len(sample)/2, len(sample)/2]
```

```python
# Perform Chi-Square Test for Goodness-of-Fit
chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)
print(f"Chi-Square Statistic: {chi2_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Fail to Reject Null Hypothesis')
```

**Observations**:
- Since the p-value < alpha, we reject $H_0$ and conclude the proportions are not equally likely.

#### 2. Kolmogorov-Smirnov Test
- How to Calculate: Use `scipy.stats.kstest(data, 'distribution')`.
- Null Hypothesis ($H_0$): The sample follows a normal distribution.
- Alternative Hypothesis ($H_a$): The sample does not follow a normal distribution.

```python
import numpy as np

data = tips['total_bill']
KS_stat, p_value = stats.kstest(data, 'norm', args=(data.mean(), data.std()))
print(f"KS Statistic: {KS_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Fail to Reject Null Hypothesis')
```

**Observations**:
- Since the p-value < alpha, we reject $H_0$ and conclude the total bill does not follow a normal distribution.

#### 3. One-Sample t-test
- How to Calculate: Use `scipy.stats.ttest_1samp(data, popmean)`.


Example: The average total bill in the "tips" dataset is ₹20.
- Null Hypothesis ($H_0$): The average total bill is ₹20.
- Alternative Hypothesis ($H_a$): The average total bill is not ₹20.

```python
t_stat, p_value = stats.ttest_1samp(tips['total_bill'], 20)
print(f"T-Statistic: {t_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Accept Null Hypothesis')
```

**Observations**:
- Since the p-value > alpha, we accpet $H_0$ and conclude the average total bill is ₹20 (approx)

#### 4. Pearson Correlation Test
- How to Calculate: Use `scipy.stats.pearsonr(x, y)`.

Example: Total bill and tip amount in the "tips" dataset are linearly related.
- Null Hypothesis ($H_0$): There is no linear relationship between total bill and tip amount.
- Alternative Hypothesis ($H_a$): There is a linear relationship between total bill and tip amount.

```python
p_stat, p_value = stats.pearsonr(tips['total_bill'], tips['tip'])
print(f"Pearson Correlation Coefficient: {p_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Accept Null Hypothesis')
```

**Observations**:
- Since the p-value < alpha, we reject $H_0$ and conclude that total bill and tip are not linearly correlated.

#### 5. Chi-Square Test of Independence
- How to Calculate: Use `scipy.stats.chi2_contingency(table)`.

Example: Gender and smoking habits in the "tips" dataset are independent.
- Null Hypothesis ($H_0$): Gender and smoking habits are independent.
- Alternative Hypothesis ($H_a$): Gender and smoking habits are dependent.

```python
import pandas as pd
```

```python
contingency_table = pd.crosstab(tips['sex'], tips['smoker'])
contingency_table
```

```python
chi2_stat, p_value, x, y = stats.chi2_contingency(contingency_table)
print(f"Chi-Square Statistic: {chi2_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Accept Null Hypothesis')
```

**Observations**:
- Since the p-value > alpha, we accept $H_0$ and conclude that the gender and smoking habits are independent.

#### 6. Two-Sample t-test
- How to Calculate: Use `scipy.stats.ttest_ind(x, y)`.

Example: The average total bill differs between smokers and non-smokers.
- Null Hypothesis ($H_0$): Smokers and non-smokers have the same average total bill.
- Alternative Hypothesis ($H_a$): Smokers and non-smokers have different average total bills.

```python
smokers = tips[tips['smoker'] == 'Yes']['total_bill']
non_smokers = tips[tips['smoker'] == 'No']['total_bill']

t_stat, p_value = stats.ttest_ind(smokers, non_smokers)
print(f"T-Statistic: {t_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Accept Null Hypothesis')
```

**Observation**:
- Since the p-value > alpha, we accept $H_0$ and conclude that the average total bill for smokers and non-smokers is the same.

#### 7. One-Way ANOVA
- How to Calculate: Use `scipy.stats.f_oneway(x1, x2, ...)`.

Example: Average total bill differs by day of the week.
- Null Hypothesis ($H_0$): Average total bill for groups is the same across all days.
- Alternative Hypothesis ($H_a$): At least one group avg bill is different.

```python
tips['day'].value_counts()
```

```python
groups = [tips[tips['day'] == day]['total_bill'] for day in tips['day'].unique()]
oneWay_stat, p_value = stats.f_oneway(*groups)
print(f"F-Statistic: {oneWay_stat}, P-Value: {p_value}")
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject Null Hypothesis')
else:
    print('Accept Null Hypothesis')
```

**Observation:**
- Since the p-value < alpha, we reject $H_0$ and conclude that average total bill differs by day.
