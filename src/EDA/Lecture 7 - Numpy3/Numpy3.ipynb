#### Comparing Arrays
Functions like np.minimum and np.maximum compare two arrays element-wise, propagating NaNs, while np.fmin and np.fmax ignore NaNs.

```python
# consider two arrays
a1 = np.array([1, -5, 6, 2])
a2 = np.array([0, np.nan, -1, -1])
```

```python
a1
```

```python
a2
```

```python
print("Ignoring NaNs (fmin): ", np.fmin(a1, a2))
print("Propagating NaNs (minimum): ", np.minimum(a1, a2))
```

```python
print("Ignoring NaNs (fmax): ", np.fmax(a1, a2))
print("Propagating NaNs (maximum): ", np.maximum(a1, a2))
```

#### Shallow Copy and Deep Copy¶

```python
x = np.array([1,2,3,4,5])
x
```

```python
y = x[:3]
y
```

```python
y[0]=1000
y
```

```python
x
```

Making changes in array y changes the element in array x too. This is known as shallow copy.

np.shares_memory(array1, array2) is used to check whether the arrays share same memory location or not

```python
np.shares_memory(x,y)
```

Therefore, it can be seen that slicing creates a shallow copy.

```python
a = np.arange(1,6)
a
```

```python
b = a+10
b
```

```python
a[0] = 9999
a
```

```python
b
```

```python
np.shares_memory(a,b)
```

This is a deep copy.

#### View vs. Copy
- A view does not create new data.
- It only creates another variable that points to the same memory.

```python
arr = np.array([10,20,30,40])
arr
```

Creating a view

```python
arr_view = arr.view()
arr_view
```

```python
arr_view[0] = 100
```

```python
arr_view
```

```python
arr
```

Changing the view also changed the original array

```python
np.shares_memory(arr, arr_view)
```

Creating a copy

```python
arr = np.array([10,20,30,40])
arr
```

```python
arr_copy = arr.copy()
arr_copy
```

```python
arr_copy[0] = 99999
```

```python
arr_copy
```

```python
arr
```

The original array remains unchanged.

```python
np.shares_memory(arr, arr_copy)
```

use of .base()

```python
print(arr_copy.base)
```

```python
print(arr_view.base)
```

.base() returns:
- None → The array owns its own memory.
- Another array → This array is a view of that array.

```python
np.shares_memory(arr_view, arr)
```

#### Ravel() vs. Flatten()

Ravel() method makes a shallow copy whereas Flatten() method makes a permanent copy after converting arrays into 1D

```python
arr1 = np.array([10,20,30,40,50,60,70,80,90]).reshape(3,-1)
```

```python
arr1
```

```python
arr1.shape
```

```python
arr_flatten = arr1.flatten()
arr_flatten
```

```python
arr_flatten.shape
```

```python
arr_ravel = arr1.ravel()
arr_ravel
```

```python
arr1
```

```python
arr_ravel.shape
```

```python
np.shares_memory(arr1, arr_flatten)
```

```python
np.shares_memory(arr1, arr_ravel)
```

```python
# item assignment to ravel array
arr_ravel[0] = 10000
```

```python
arr_ravel
```

```python
arr1
```

```python
arr_flatten
```

```python
# item assignment to flatten array
arr_flatten[2] = 99999
```

```python
arr_flatten
```

```python
arr1
```

#### Vectorization

Vectorization is a method of performing array operations without the use of for loops.

```python
arr = np.array([1,2,3,4,5,6])
arr
```

```python
def even(n):
    return n%2==0
```

```python
Even_Num = np.vectorize(even)
```

```python
Even_Num(arr)
```
