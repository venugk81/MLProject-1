## Pandas - Python library used for working with dataframes

What is Pandas?

Pandas stands for Panel data or Python Data Analysis. It is a library used for data analysis and manipulation. It creates a dataframe that is nothing but data in tabular form. Pandas works with heterogenous data. A single column in pandas is called as series. Series in pandas is like a vector in numpy (single col)

What is a dataframe?

A DataFrame is a data structure that organizes data into a 2-dimensional table of rows and columns, much like a spreadsheet
