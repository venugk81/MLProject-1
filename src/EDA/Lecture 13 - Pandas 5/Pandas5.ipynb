#### Pivoting and Melting
- Pivot converts rows → columns
- Melt converts columns → rows

Let's create a DataFrame

```python
data = {
        'Region': ['North', 'South', 'North', 'South', 'North', 'South'],
        'Product': ['A', 'A', 'B', 'B', 'C', 'C'],
        'Sales': [100, 200, 150, 300, 250, 400]
       }
df = pd.DataFrame(data)
```

```python
df
```

To see the sales of each product in every region we use pivot method here.

```python
df.pivot(index='Region', columns='Product', values='Sales')
```

- index - this decides what becomes rows
- columns - this decides what becomes the column headings
- values - this fills the table with values

The above result can be achieved using groupby() too, but pivot() is shorter and is designed specifically for reshaping the data

Ques. Find the total sales of each product.

```python
pivot_df = df.pivot(index='Region', columns='Product', values='Sales')
```

```python
pivot_df.sum()
```

Ques. Find the total sales of each region.

```python
pivot_df.sum(axis=1)
```

Ques. Which product generated the highest total sales?

```python
pivot_df.sum().idxmax()        # idxmax() - it returns the label (index) where the maximum value is found
```

Ques. Which region generated the highest total sales?

```python
pivot_df.sum(axis=1).idxmax()
```

Ques. Calculate the grand total sales of all regions and all products.

```python
pivot_df.sum().sum()
```

##### Melt Method

```python
pivot_df
```

```python
pivot_df.reset_index()
```

```python
pivot_df_reset = pivot_df.reset_index()
```

```python
pivot_df_reset.melt(id_vars='Region', var_name='Product', value_name='Sales')
```

- As region is not a column and melt works with columns, therefore resetting the index first.
- id_vars: These columns remain unchanged and are repeated for every melted row.
- var_name: New column storing old column names.
- value_name: The new column that stores the values.

#### Stack and Unstack
- Stack is also used to convert columns into rows but it moves existing column labels into the row index.
- It produces a MultiIndex Series/DataFrame
- Once the data is stacked, it becomes a Series with a MultiIndex.

```python
df.head()
```

```python
stacked_df = df.stack()
```

```python
stacked_df
```

```python
type(stacked_df)
```

```python
stacked_df.index
```

There are 2 index levels here:
- Level 0 → 0,1,2,3,4,5
- Level 1 → Region, Product, Sales

Ques. Find how many Region values are present.

```python
stacked_df.xs('Region', level=1)
```

df.xs → It stands for Cross Section and is used to extract (select) data from a particular level of a MultiIndex.

Ques. Find all Product values.

```python
stacked_df.xs('Product', level=1)
```

Ques. Find all Sales values.

```python
stacked_df.xs('Sales', level=1)
```

Ques. Find the maximum sales value.

```python
stacked_df.xs('Sales', level=1).max()
```

Ques. Find the count of each region.

```python
stacked_df.xs('Region', level=1).value_counts()
```

#### Unstack Data

```python
stacked_df.unstack()
```
